#Valid Anagram

1. Understanding & Breakdown

This problem asks you to determine if two strings are anagrams, meaning they consist of identical characters, each appearing exactly once, but in different orders. The input is two strings, s and t, and the output is a boolean value indicating whether t is a valid anagram of s. I drew the following flowchart.

2. Peer Review & Draft Code

I reviewed a solution under the LeetCode and I find a code example by using Python:

https://leetcode.com/problems/valid-anagram/solutions/6101148/video-4-solutions-by-niits-x6b2/

His approach is to use a hash table to count the frequency of characters in the first string, and then use the second string to cancel out these counts one by one. If the count is insufficient or the character does not exist, he returns False. If all the characters are successfully canceled out, it means that the two strings are anagrams. I tried to write a draft code myself and referred to his logic:

```python
def validAnagram(s: str, t: str) -> bool:
  if len(s) != len(t):
    return False
  count = {}
  for char in s:
    count[char] = count.get(char, 0) + 1
  for char in t:
    if char not in count:
      return False
    count[char] -= 1
    if count[char] < 0:
      return False
  return True
```
